## 2. Importar y revisar el CSV original

```python
from manejador_csv import ManejadorCSV

csv_handler = ManejadorCSV(CSV_ORIGINAL)
encabezados_csv, registros_csv = csv_handler.importar()

print("Encabezados CSV:", encabezados_csv)
print("Cantidad de registros CSV:", len(registros_csv))
print("Primeros 3 registros CSV:")
for fila in registros_csv[:150]:
    print(fila)
```

## 3. Modificar el CSV y guardar una nueva versión

```python
# Modificación de demostración: actualizar el valor de la especie en el CSV.
# Si el archivo descargado usa otros valores, ajusta estos dos argumentos.
columna_especie = "Species"
valor_anterior = "Iris-setosa"
valor_nuevo = "Iris-setosa (revisado)"

if columna_especie in csv_handler.encabezados:
    cambios_csv = csv_handler.actualizar_valor(
        columna_especie, valor_anterior, valor_nuevo
    )
    print("Cambios CSV realizados:", cambios_csv)
else:
    print(f"No existe la columna {columna_especie}. Ajusta la modificación a las columnas reales del CSV.")

ruta_csv_guardado = csv_handler.guardar(CSV_MODIFICADO)
print("CSV modificado guardado en:", ruta_csv_guardado)
print("Vista posterior (primeros 3 registros):")
for fila in csv_handler.registros[:3]:
    print(fila)
```

## 5. Modificar el JSON y guardar una nueva versión

```python

# Modificar un registro del archivo JSON y guardar una nueva versión.
# Se conserva el archivo original sin cambios.

if datos_json:
    campo_existente = "sentence1"

    valor_anterior, valor_nuevo = json_handler.actualizar_valor(
        0,
        campo_existente,
        datos_json[0][campo_existente] + " [revisado]"
    )

    print(f"Campo modificado: {campo_existente}")
    print("Antes:", valor_anterior)
    print("Después:", valor_nuevo)

    ruta_json_guardado = json_handler.guardar(JSON_MODIFICADO)

    print("JSON modificado guardado en:", ruta_json_guardado)
    print("Primer registro después de la modificación:")
    print(json_handler.datos[0])
else:
    print("No hay registros JSON para modificar.")
```

## 6. Generar el resumen TXT con resultados reales

```python

# Actualizar los objetos con las clases recargadas
import importlib
import manejador_csv
import manejador_json
from generador_resumen_txt import GeneradorResumenTXT

importlib.reload(manejador_csv)
importlib.reload(manejador_json)

ManejadorCSV = manejador_csv.ManejadorCSV
ManejadorJSON = manejador_json.ManejadorJSON

# Leer nuevamente los archivos originales
csv_handler = ManejadorCSV(CSV_ORIGINAL)
encabezados_csv, registros_csv = csv_handler.importar()

json_handler = ManejadorJSON(JSON_ORIGINAL)
datos_json = json_handler.importar()

# Generar el resumen con los objetos actualizados
resumen = GeneradorResumenTXT(
    url_csv=URL_CSV_ORIGEN,
    url_json=URL_JSON_ORIGEN,
    csv_original=str(CSV_ORIGINAL),
    csv_salida=str(CSV_MODIFICADO),
    json_original=str(JSON_ORIGINAL),
    json_salida=str(JSON_MODIFICADO),
    manejador_csv=csv_handler,
    manejador_json=json_handler,
)

ruta_txt = resumen.generar(TXT_RESUMEN)

print("Registros CSV:", csv_handler.cantidad_registros)
print("Registros JSON:", len(datos_json))
print("Resumen generado en:", ruta_txt)
print("\n" + Path(ruta_txt).read_text(encoding="utf-8"))
```
